## Exercícios 03 - Estruturas Repetitivas

Exercício 01

```python
print("Digite dois numeros:")
x = input()
y = input()

while x != y:
    if x > y:
        print("Decrescente")
        print("Digite outros dois numeros:")
        x = input()
        y = input()
    if x < y:
        print("Crescente")
        print("Digite outros dois numeros:")
        x = input()
        y = input()
```

Exercicio 02

```python
soma = 0
count = 0
    
print("Digite as idades (digite um número negativo para finalizar):")
    
while True:
    idade = int(input())
    
    if count == 0 and idade < 0:
        print("Impossível calcular!")
    
    if idade < 0:
        break
        
    soma += idade
    count += 1

if count > 0:
    media = soma / count
    print(f"Média = {media:.2f}")
else:
    print("Impossível calcular")
```
