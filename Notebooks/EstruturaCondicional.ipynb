Exercício 02

```python
import math as m

a = float(input("Coeficiente a: "))
b = float(input("Coeficiente b: "))
c = float(input("Coeficiente c: "))

delta = m.pow(b, 2) - 4 * a * c

if (delta > 0):
    x1 = (-b + m.sqrt(delta)) / 2
    x2 = (-b - m.sqrt(delta)) / 2
    print(f"X1: {x1:.4f}")
    print(f"X2: {x2:.4f}")
elif (delta == 0):
    x1 = (-b + m.sqrt(delta)) / 2
    print(f"X1 = X2 = X: {x1:.4f}")
else:
    print("Esta equação não possui raízes reais!")
```

Exercício 03

```python
num1 = int(input("Primeiro valor: "))
num2 = int(input("Segundo valor: "))
num3 = int(input("Terceiro valor: "))

menor = min(num1, num2, num3)

if (num1 == num2 == num3):
    print(f"Menor: {menor}")
elif num1 == num2 and num1 != num3:
    print(f"Menor: {menor}")
elif num2 == num3 and num2 != num1:
    print(f"Menor: {menor}")
elif num1 == num3 and num1 != num2:
    print(f"Menor: {menor}")
else:
    print(f"Menor: {menor}")
```

Exercício 04

```python
min = int(input("Digite a quantidade de minutos: "))

if (min > 100):
    minExc = min - 100
    valExc = 2 * minExc
    print(f"Valor a pagar: R$ {(50 + valExc):.2f}")
else:
    print("Valor a pagar: R$ 50.00")
```

Exercício 05

```python
valUnit = float(input("Preço unitário do produto: "))
qtde = int(input("Quantidade comprada: "))
valReceb = float(input("Dinheiro recebido: "))

valTotal = qtde * valUnit

if (valReceb > valTotal):
    print(f"Troco: R$ {(valReceb - valTotal):.2f}")
elif (valReceb == valTotal):
    print("Troco: R$ 0.00")
elif (valReceb < valTotal):
    print(f"Valor insuficiente. Faltam R$ {((valReceb - valTotal) * (-1)):.2f}")
```

Exercício 06

```python
gli = float(input("Entre com o valor de glicose do paciente (mg/dL): "))

if (gli > 140):
    print("Classificação: Diabetes")

if (gli > 100 and gli <= 140):
    print("Classificação: Elevado")

if (gli <= 100):
    print("Classificação: Normal")
```
